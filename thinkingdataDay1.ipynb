```python
import json
```

```python
#function to load the data
def load_data(filename):
    with open(filename, "r")as f:
        data = json.load(f)
    return data
```

```python
data = load_data("storedata.json")
print(data)
print(type(data))
```

```python
# Clean & structure the data
def clean_data(data):
    text_to_num = {
        "one": 1,
        "two": 2,
        "three": 3,
        "four": 4,
        "five": 5
    }

    unique_users = set()
    cleaned_data = []

    for user in data:
        # Clean rating - consistent typing 
        #strip() # Removes extra spaces from the beginning and end of a string
        #.lower() # Converts all letters in a string to lowercase
        raw_rating = str(user["rating"]).strip().lower()  


        if raw_rating in text_to_num:
            raw_rating = text_to_num[raw_rating]

        user["rating"] = raw_rating

        # Handle missing values
        #.get() # Gets a value from a dictionary safely
        raw_age = user.get("age")

        if raw_age is None:
            user["age"] = None

        # Handle duplicate values - deduplication
        #.append()     # Adds an item to the end of a list
        name = user["name"].strip()

        if name in unique_users:
            continue

        unique_users.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data
```

```python
clean_data(data)
```

```python
#Finding meaningful insights fome data like  AVG Rating and percenage of rating less
def get_insights(data):
    #avg rating for data of product
    tot_ratings = 0
    for user in data:
        tot_ratings += float(user['rating'])
    
    print(f"average product rating = {tot_ratings/len(data)}")

    #percentage users with poor rating for product
    poor_ratings = 0
    for user in data:
        if(float(user['rating']) < 3.0):
            poor_ratings += 1
    
    print(f"% users with poor rating = {poor_ratings/len(data)*100}%")
```

```python
get_insights(data)
```

```python
#product recommendation feature
#1. If user rating >= 4, recommend same brand products.
#2. If user rating < 4, recommend different brand products.
def get_recommendations(data):
    recommendations = []
    
    for user in data:
        curr_recomm = {}
        curr_recomm['name'] = user['name']
        if(float(user['rating']) >= 4):
            curr_recomm['brand'] = "Apple"
        else:
            curr_recomm['brand'] = "Samsung"

        recommendations.append(curr_recomm)
    
    return recommendations
```

```python
recommendations = get_recommendations(data)
print(recommendations)
```
